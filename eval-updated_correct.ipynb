```python
import numpy as np
import pandas as pd 
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Salary_Data.csv")
print(df.head())
print(df.columns)

df.isnull().sum()
```

```python
# Udpated ---- 

df["Salary"]=df["Salary"].fillna(df["Salary"].mean())
df["Years of Experience"] = df["Years of Experience"].fillna(df["Years of Experience"].mean())

df.isnull().sum()

X = np.array(df["Years of Experience"]).reshape(-1,1)
y = np.array(df["Salary"]).reshape(-1,1)
print(X,y,sep="-------")
```

```python
# Updated 
plt.scatter(X,y)
```

```python
model_updated = LinearRegression()
model_updated.fit(X,y)
```

```python
model_updated.intercept_,model_updated.coef_
```

```python
X_nw = np.array([[2],[25],[3],[7],[2],[9]])
y_pred = model_updated.predict(X_nw)
print(y_pred)
```
